## 1. Daten laden

```python
from sonnenfeld import modelling

# Load the cleaned table and keep only rows with a satisfaction value (MCAR, see EDA 2.4)
df = pd.read_csv(config.INPUT_FILE)
df_model = df.dropna(subset=["zufriedenheit"])

# Base feature set, identical for all models (see section 3)
merkmale = [
    "zufriedenheit",
    "abteilung",
    "betriebszugehoerigkeit_monate",
    "position",
    "ueberstunden_monat",
    "abwesenheitstage_jahr",
]

# Stratified 80/20 split (target share equal in train and test)
X_train, X_test, y_train, y_test = modelling.split(df_model, merkmale)

# Control: sizes and attrition rate in both parts
print(f"Train: {len(X_train):,} Zeilen | Test: {len(X_test):,} Zeilen")
print(f"Fluktuationsquote Train: {y_train.mean():.1%} | Test: {y_test.mean():.1%}")
```

**Erkenntnis:** Stratifizierte Aufteilung (80/20, random_state = 42): 7.600 Trainings- und
1.900 Testdatensätze; die Fluktuationsquote ist in beiden Teilen identisch (16,9 %).

## 3. Features festlegen

*Warum dieser Schritt?* …

**Basismodell (für alle Modelle identisch):** `zufriedenheit`, `abteilung`,
`betriebszugehoerigkeit_monate`, `position`, `ueberstunden_monat`, `abwesenheitstage_jahr`

**Bewusst ausgeschlossen:**
- `mitarbeiter_id`: Kennung ohne Informationsgehalt
- `geschlecht`: ethische und rechtliche Gründe (AGG); nur für die Fairness-Prüfung
- `alter`: kein Zusammenhang mit der Fluktuation (H1c, p = 0,56) sowie Diskriminierungsrisiko (AGG)
- `eintrittsdatum`, `betriebszugehoerigkeit_jahre`: redundant zur Betriebszugehörigkeit in Monaten
- `gehalt_brutto`, `gehalt_status`: Rohwerte bzw. Metadaten der Bereinigung
- `gehalt_final`: stark von der Hierarchieebene abhängig (Multikollinearität); `position` ist
  besser interpretierbar → Test als Variante
- `gehalt_rang`: kein Zusammenhang mit der Fluktuation (H3b)
- Gruppenvariablen (`_gruppe`): Originalwerte enthalten mehr Information

**Geplante Varianten:** ohne `abwesenheitstage_jahr` (Sensitivitätsanalyse),
mit `gehalt_final`, mit binären Schwellenwerten für die logistische Regression.
